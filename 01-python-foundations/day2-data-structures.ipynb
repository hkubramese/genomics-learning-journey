## FASTQ File StructureFASTQ files store the raw read data coming out of DNA/RNA sequencers.Each read consists of 4 lines:1. Header line2. Sequence line3. Plus line4. Quality lineExample:```@read_1ATGCGTAC+IIIIIIII```Here:- `@read_1` — the read ID.- `ATGCGTAC` — the DNA sequence.- `+` — separates the sequence from the quality info.- `IIIIIIII` — represents the quality score for each base.

````python
fastq_text = """@read_1ATGCGTAC+IIIIIIII@read_2GGGTTTAA+HHHHHHHH@read_3CCATGNTA+FFFFF#FF"""with open("example.fastq", "w") as f:    f.write(fastq_text)print("example.fastq file created.")
````

## Manual FASTQ ParsingParsing means pulling meaningful information out of a file.For FASTQ, our goal is to extract:- the read ID- the sequence- the quality info- and turn all of that into something we can analyze

````python
with open("example.fastq", "r") as f:    while True:        header = f.readline().strip()        if not header:            break        sequence = f.readline().strip()        plus = f.readline().strip()        quality = f.readline().strip()        print("Header:", header)        print("Sequence:", sequence)        print("Quality:", quality)        print("Read length:", len(sequence))        print("-" * 30)
````

## Calculating GC Content from FASTQGC content is the proportion of G and C bases in a DNA sequence.Formula:GC content = (G count + C count) / total base count

````python
with open("example.fastq", "r") as f:    while True:        header = f.readline().strip()        if not header:            break        sequence = f.readline().strip()        plus = f.readline().strip()        quality = f.readline().strip()        gc = (sequence.count("G") + sequence.count("C")) / len(sequence)        print(header, "GC content:", gc)
````

## Let's Write a Parser FunctionWe can turn the FASTQ-reading logic into a function to cut down on repeated code.

````python
def parse_fastq(filename):    records = []    with open(filename, "r") as f:        while True:            header = f.readline().strip()            if not header:                break            sequence = f.readline().strip()            plus = f.readline().strip()            quality = f.readline().strip()            record = {                "header": header,                "sequence": sequence,                "quality": quality            }            records.append(record)    return records
````

````python
records = parse_fastq("example.fastq")for record in records:    print(record)
````

````python
# Summary statistics using manual parsingrecords = parse_fastq("example.fastq")lengths = []gc_values = []for record in records:    seq = record["sequence"]    lengths.append(len(seq))    gc = (seq.count("G") + seq.count("C")) / len(seq)    gc_values.append(gc)print("Number of reads:", len(records))print("Average read length:", sum(lengths) / len(lengths))print("Average GC content:", sum(gc_values) / len(gc_values))
````

## Limits of Manual ParsingManual parsing is great for learning, because it forces you to understand the file format.But in real projects, it's risky:- A small quirk in the file format can break the code.- It doesn't automatically convert quality scores.- Memory usage can become a problem on large files.- The code gets longer, and the risk of bugs grows.That's why real bioinformatics analyses usually rely on libraries like Biopython instead.

## What Is Biopython?Biopython is a Python library for working with biological data.It can read many formats — FASTA, FASTQ, GenBank, and more.The `SeqIO` module is used to read and write sequence files.

## Quality ScoresQuality info in a FASTQ file is stored as characters.Biopython automatically converts these characters into Phred quality scores.Higher score = more reliable base callLower score = less confidence

````python
for record in SeqIO.parse("example.fastq", "fastq"):    qualities = record.letter_annotations["phred_quality"]    print(record.id)    print("Quality scores:", qualities)    print("Average quality:", sum(qualities) / len(qualities))    print("-" * 30)
````

This is where we see the biggest difference between FASTQ and FASTA. FASTQ doesn't just store the sequence — it also stores how reliable that sequence is.

## Filtering Out Low-Quality ReadsIn real analyses, low-quality reads are usually filtered out.For example, we might keep only reads with an average quality score above 30.

````python
high_quality_reads = []for record in SeqIO.parse("example.fastq", "fastq"):    qualities = record.letter_annotations["phred_quality"]    avg_quality = sum(qualities) / len(qualities)    if avg_quality >= 30:        high_quality_reads.append(record)print("Number of high-quality reads:", len(high_quality_reads))
````
