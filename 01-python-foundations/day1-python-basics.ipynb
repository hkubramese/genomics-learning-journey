**2. Conditional statements (If - Elif - Else):**

**General syntax**In Python, the structure that lets the program make decisions is the `if-elif-else` block. Python reads these conditions in order, top to bottom.General skeleton:```python# --- IF-ELIF-ELSE SKELETON ---condition_1 = Truecondition_2 = Falseif condition_1:    # Runs only if condition_1 is True.    # Code inside must be indented (Tab or 4 spaces)!    print("First block ran.")elif condition_2:    # Runs if condition_1 is False but condition_2 is True.    # You can chain as many 'elif' blocks as you need.    print("Second block ran.")else:    # Runs only if none of the above conditions were True.    # NOTE: 'else' never takes a condition!    print("Last block (else) ran.")```**Basic rules:**- **Colon (`:`)** — every `if`, `elif`, and `else` line must end with `:`.- **Indentation** — Python doesn't use curly braces `{}` like some other languages. Leading whitespace (indentation) is what tells Python which code belongs to which condition. Forget it and you get an `IndentationError`.

## First hands-on task (reading a FASTA file manually, length and GC content)

**What is a FASTA file?**Biological sequences (DNA, RNA, protein) are very long. Instead of storing them in something complex like Word or Excel, we use a very simple, universal plain-text format for them — FASTA.The FASTA format has only two golden rules:- **ID line:** always starts with a greater-than sign (`>`). This line holds the gene name, organism info, or database accession.- **Sequence lines:** start right after the ID line. They only contain biological letters (A, T, G, C, or amino acids). For readability, the sequence is usually wrapped every 80 characters.Example FASTA record:```text>NM_000546.6 Homo sapiens tumor protein p53 (TP53), mRNAGATCGATCGATCGATCGATCGATCGATCGATCGATCCGATCGATCGATCGATCGATCGATCGATCGATCGAT```

**Step 1: writing a FASTA file**Before we can read a file, we need one to read — let's use Python to write a small FASTA file onto Colab's virtual disk.

````python
# Writing a FASTA file into Colab using Pythonfasta_text = """>seq1_sample_sequenceATGCGTACGTAGCTAGCTAGCTGATGCTAGCTGATCCGTAGCTGATCGTAGCTAGCTGATCGTAGCTGATCG"""with open("sample.fasta", "w") as file:    file.write(fasta_text)print("sample.fasta created successfully! You can check it in the file browser on the left.")
````

**Step 2: reading the FASTA file manually**Now let's open it and write a program that stores the header separately from the sequence. This is exactly where the `for` loop and `if-else` conditions we just learned come in!

````python
# Setting up empty variables to hold the datasequence_id = ""dna_sequence = ""# Opening the file in read mode ('r')with open("sample.fasta", "r") as file:    for line in file:        line = line.strip()  # removes invisible trailing whitespace/newlines (\n)        if line.startswith(">"):            # This is the header line            sequence_id = line[1:]  # start from index 1 to drop the leading '>'        else:            # This is a DNA sequence line, append it to what we have so far            dna_sequence = dna_sequence + lineprint("Sequence ID:", sequence_id)print("Combined DNA sequence:", dna_sequence)
````

**Step 3: computing length and GC content**We've read the file and have our sequence in hand. So what next? In bioinformatics, there are two basic things we check first when we get a sequence:- **Sequence length** — to confirm the gene/sequence we read is the size (in bp) we expected.- **GC content** — Guanine (G) and Cytosine (C) pair with 3 hydrogen bonds, while Adenine (A) and Thymine (T) pair with only 2. So sequences with higher GC content are more stable and need higher temperatures to denature (separate by heating). This ratio is especially critical when designing PCR primers.Let's calculate the length and GC ratio of the sequence we just read.

````python
# 1. Total lengthtotal_length = len(dna_sequence)# 2. Counting G and C (using .upper() first is good practice to avoid case issues)dna_sequence = dna_sequence.upper()g_count = dna_sequence.count("G")c_count = dna_sequence.count("C")# 3. Calculating GC ratio# Formula: ((G + C) / total length) * 100gc_ratio = ((g_count + c_count) / total_length) * 100print("--- ANALYSIS RESULTS ---")print("Sequence length:", total_length, "bp")print("Guanine (G) count:", g_count)print("Cytosine (C) count:", c_count)print("GC content: %", round(gc_ratio, 2))  # round() to show 2 decimal places
````
