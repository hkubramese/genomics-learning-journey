## Hands-On Task — Reading and Summarizing a VCF File**Goal:** understand VCF, one of the most critical file formats in bioinformatics, and use Pandas to filter a huge mutation dataset down to high-quality variants.#### 1. What Is a VCF (Variant Call Format) File?The human genome has about 3.2 billion letters, but the difference between any two people is only around 0.1% (roughly 3-4 million letters/mutations). Instead of storing the whole genome, we store just the points that DIFFER from the reference genome (mutations, SNPs) — and that table format is called **VCF**.**Core VCF columns:*** **CHROM:** chromosome number* **POS:** the mutation's position in the genome* **REF:** the reference (normal) base* **ALT:** the alternate (mutant) base* **QUAL:** sequencing quality score (higher is better)* **FILTER:** did it pass quality checks? (PASS)#### 2. Creating a Sample VCF File in ColabLet's create a small VCF file to analyze. Real VCF files start with hundreds of metadata lines beginning with `##`; we're just using the table portion here to keep things simple.

```python
# Creating a tab-separated VCF table (\t = tab character)vcf_text = """#CHROM\tPOS\tID\tREF\tALT\tQUAL\tFILTER\tINFOchr1\t10020\trs123\tA\tG\t95.5\tPASS\tDP=120chr1\t10045\trs456\tT\tC\t12.0\tFAIL\tDP=10chr2\t50030\trs789\tG\tT\t150.2\tPASS\tDP=300chr2\t50099\trs111\tC\tA\t8.5\tFAIL\tDP=5chr3\t80000\trs222\tA\tT\t65.0\tPASS\tDP=85chr3\t80500\trs333\tG\tC\t99.9\tPASS\tDP=150"""with open("mutations.vcf", "w") as file:    file.write(vcf_text)print("mutations.vcf file created successfully!")
```

#### 3. Analyzing the VCF File with PandasNow let's read this table and filter down to only **reliable** mutations (those with `PASS` and `QUAL > 50`).

```python
# 1. Reading the file (using sep='\t' since it's tab-separated, not comma-separated)vcf_df = pd.read_csv("mutations.vcf", sep="\t")print("--- RAW VCF DATA ---")display(vcf_df)
```

```python
# 2. Filtering out low-quality mutations# Criterion 1: the FILTER column must be 'PASS'# Criterion 2: the QUAL (quality) score must be greater than 50reliable_variants = vcf_df[(vcf_df['FILTER'] == 'PASS') & (vcf_df['QUAL'] > 50)]print("\n--- HIGH-QUALITY (RELIABLE) VARIANTS ---")display(reliable_variants)
```

```python
# 3. Extracting a statistic: how many high-quality mutations per chromosome?chrom_counts = reliable_variants['#CHROM'].value_counts().reset_index()chrom_counts.columns = ['Chromosome', 'Mutation_Count']print("\n--- MUTATION DISTRIBUTION BY CHROMOSOME ---")display(chrom_counts)
```

Instead, they use Linux terminal commands. In Colab, putting an exclamation mark (`!`) in front of a line makes it run as a Linux command instead of Python!Let's download all the mutations for chromosome 22 from the 1000 Genomes Project servers (about 12 MB compressed, 1M+ rows) using **`!wget`**.

```python
import gzipimport osprint("Downloading 1000 Genomes Project data...")url = "https://hgdownload.cse.ucsc.edu/gbdb/hg19/1000Genomes/phase3/ALL.chr22.phase3_shapeit2_mvncall_integrated_v5a.20130502.genotypes.vcf.gz"!wget --show-progress "$url" -O chr22_real.vcf.gz
```

**5. Reading a Huge File While Skipping the Junk Lines (Data Engineering)**Real VCF files start with hundreds of `##` (metadata) lines describing how the file was generated. The actual column headers are on the single-`#` line, `#CHROM`. We need to read this intelligently with Python.

```python
import gzipimport pandas as pd# Find the header linewith gzip.open("chr22_real.vcf.gz", "rt") as f:    for line in f:        if line.startswith("#CHROM"):            headers = line.strip().split("\t")            break# Let's read only the columns we actually needusecols = ["#CHROM", "POS", "ID", "REF", "ALT", "INFO"]vcf_df = pd.read_csv(    "chr22_real.vcf.gz",    sep="\t",    comment="#",    header=None,    names=headers,    usecols=usecols)print("VCF file loaded")print(f"Total number of variants: {vcf_df.shape[0]:,}")vcf_df.head(3)
```

**6. Data Mining: Identifying SNPs and Parsing the INFO Column**At this scale — well beyond what Excel could even open — the INFO column is a mess. What we need out of it is how common each mutation is in the population (AF: Allele Frequency).

```python
# A function to pull the AF value out of the INFO columndef get_af(info):    for item in info.split(";"):        if item.startswith("AF="):            return item.replace("AF=", "")    return None# Creating the AF columnvcf_df["AF"] = vcf_df["INFO"].apply(get_af)# Converting AF to a numeric typevcf_df["AF"] = pd.to_numeric(vcf_df["AF"], errors="coerce")vcf_df[["#CHROM", "POS", "ID", "REF", "ALT", "AF"]].head()
```

```python
# SNP (Single Nucleotide Polymorphism)snp_df = vcf_df[    (vcf_df["REF"].str.len() == 1) &    (vcf_df["ALT"].str.len() == 1)]print(f"Total number of SNPs: {snp_df.shape[0]:,}")snp_df[["#CHROM", "POS", "ID", "REF", "ALT", "AF"]].head()
```

**7. Case Study: Filtering for Rare Disease/Cancer Mutations**You've got over a million variants on chromosome 22. Most of them are just normal differences that make you "you" — hair color, height, and so on. A researcher looking for rare disease genes like cancer focuses on mutations seen in less than 1% of the population (`AF < 0.01`).

```python
rare_df = vcf_df[vcf_df["AF"] < 0.01]print(f"Number of rare variants: {rare_df.shape[0]:,}")rare_df[["#CHROM", "POS", "REF", "ALT", "AF"]].head()
```

```python
rare_snp_df = vcf_df[    (vcf_df["AF"] < 0.01) &    (vcf_df["REF"].str.len() == 1) &    (vcf_df["ALT"].str.len() == 1)]print(f"Number of rare SNPs: {rare_snp_df.shape[0]:,}")
```

---## Getting to Know Numpy — Genomic Matrices**Goal:** get familiar with Numpy, the engine behind Pandas, and understand where numeric arrays and matrices come in — especially for machine learning and genetic distance calculations.#### 1. What Is Numpy, and Why Do We Need It?Regular Python lists are too slow for heavy math. **Numpy (Numerical Python)** organizes numbers into tables (matrices) and lets us run math operations across them at very high speed.

```python
# Importing Numpy under the alias 'np'import numpy as np# Expression levels for 3 patients (A, B, C) across 4 genes (G1, G2, G3, G4)# This is a 3x4 matrix (a 2D array)expression_matrix = np.array([    [10, 15, 12, 11],     # Patient A    [5,  5,  6,  4],      # Patient B (genes don't look very active)    [100, 120, 130, 110]  # Patient C (genes are extremely active — maybe a tumor cell)])print("--- NUMPY MATRIX ---")print(expression_matrix)
```

```python
print(f"\nMatrix shape (rows, columns): {expression_matrix.shape}")
```

#### 2. Fast Biological Calculations on MatricesHow do we get the average for every patient WITHOUT using a `for` loop?

```python
# 1. Finding each patient's (row's) overall average# axis=1 means "operate row-wise / horizontally"patient_averages = np.mean(expression_matrix, axis=1)print("Average gene expression per patient:")print(f"Patient A: {patient_averages[0]}")print(f"Patient B: {patient_averages[1]}")print(f"Patient C: {patient_averages[2]}")
```

```python
# 2. Finding the maximum value for each gene (column)# axis=0 means "operate column-wise / vertically"gene_maxima = np.max(expression_matrix, axis=0)print("\nMaximum observed value per gene (G1, G2, G3, G4):")print(gene_maxima)
```

```python
# 3. Normalization: doubling all values (e.g. signal amplification)# We can multiply the entire matrix at once in Numpy - no loop needed!amplified_matrix = expression_matrix * 2print("\nMatrix doubled:")print(amplified_matrix)
```

**3. Advanced Bioinformatics Operations (Log Transform and Masking)**Numpy's real power shows up when preparing biological data for machine learning.

```python
print("--- DATA TRANSFORMATION AND FILTERING ---")# 1. LOG TRANSFORM (normalizing the data)# In bioinformatics, we log2-transform data to close the huge gaps between RNA counts.# (Adding +1 avoids errors on zero values - log2(x+1) is a common bioinformatics trick)log_matrix = np.log2(expression_matrix + 1)print("1. Log2-transformed expression matrix (notice how the huge gaps are compressed):")print(np.round(log_matrix, 2))
```

```python
# 2. BOOLEAN MASKING (which genes are expressed above 50?)# Keep only values greater than 50, everything else is droppedmask = expression_matrix > 50# Selecting only the values that satisfy the mask (filtering)active_genes = expression_matrix[mask]print("\n2. Expression values greater than 50 (highly active):")print(active_genes)
```

**4. Genetic Distance and Similarity: Are the Patients Related? (Correlation)**When clustering patients, machine learning algorithms look at the distance (mathematical similarity) between them.

```python
print("--- SIMILARITY BETWEEN PATIENTS (CORRELATION) ---")# Calculating the pairwise similarity (Pearson correlation) within the patient matrix# The closer to 1, the more similar they are.similarity_matrix = np.corrcoef(expression_matrix)print("Correlation matrix (comparing patients A, B, and C to each other):")print(np.round(similarity_matrix, 3))print("\nInterpretation:")print("Similarity between Patient A and Patient B:", round(similarity_matrix[0, 1], 3), "-> (quite high!)")print("Similarity between Patient A and Patient C:", round(similarity_matrix[0, 2], 3), "-> (lower)")
```
